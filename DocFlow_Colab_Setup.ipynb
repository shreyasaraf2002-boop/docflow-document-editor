## Important limitations

- Closing or resetting the Colab runtime stops the preview. Files stored under `/content` are temporary.
- This is **not** a live deployed URL suitable for reviewers.
- The demo user switcher is mock identity, not secure authentication.
- The Render free plan's filesystem is temporary; do not claim SQLite survives service restart. For durable public hosting, connect a managed database or attach a persistent disk on a suitable paid plan.
